```python
import pandas as pd
import seaborn as sns
df = sns.load_dataset("titanic")
print("Titanic dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())
print("\nDataset Information:")
df.info()
print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
# Fill missing Age values with the median
df["age"] = df["age"].fillna(df["age"].median())
# Fill missing Embarked values with the mode
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
# Fill missing Embark Town values with the mode
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
print("Missing values handled!")
print(df.isnull().sum())
```

```python
print("Survival rate by gender:")
display(df.groupby("sex")["survived"].mean())
print("\nSurvival rate by passenger class:")
display(df.groupby("class")["survived"].mean())
print("\nAverage fare by passenger class:")
display(df.groupby("class")["fare"].mean())
```

```python
print("Key Statistical Insights:")
print("\n1. Overall survival rate:")
print(df["survived"].mean())
print("\n2. Number of passengers by gender:")
print(df["sex"].value_counts())
print("\n3. Number of passengers by class:")
print(df["class"].value_counts())
print("\n4. Average passenger age:")
print(df["age"].mean())
print("\n5. Average fare:")
print(df["fare"].mean())
```

```python
print("Final dataset shape:", df.shape)
print("\nFinal missing values:")
print(df.isnull().sum())
```
